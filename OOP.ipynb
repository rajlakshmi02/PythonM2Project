```python
class Car:
    def __init__(self, brand, color, speed):
        self.brand = brand
        self.color = color
        self.speed = speed

    def display_details(self):
        print("Brand:", self.brand)
        print("Color:", self.color)
        print("Speed:", self.speed, "km/hr")

    def check_speed(self):
        if self.speed > 100:
            print("High Speed")
        else:
            print("Normal Speed")


# Creating an object
car1 = Car("Toyota", "Red", 120)

# Calling methods
car1.display_details()
car1.check_speed()
```

```python
class Bank():
    def __init__(self, Name, accountno,initialbal=0):
        self.Name= Name
        self.accountno= accountno
        self.balance= initialbal
    def Deposit(self, amount):
        if amount>0:
            self.balance += amount
            print("the Current balance after deposit: ₹", self.balance)
        else:
            print("deposit amount cannot be negative")
    def Withdraw(self, amount):
        if 0<amount<= self.balance:
            self.balance -=amount
            print("the current balance after withdraw:₹", self.balance)
        else:
            print("insufficient fund/invalid amount")
    def get_balance(self):
        return self.balance

account=Bank("rama","12345")

account.Deposit(1500)
account.Withdraw(500)
account.get_balance()
```

```python
class Bank:
    bank_name = "SBI"
    total_accounts = 0

    def __init__(self, Name, accountno, initialbal=0):
        self.Name = Name
        self.accountno = accountno
        self.balance = initialbal
        Bank.total_accounts += 1

    @classmethod
    def show_total_accounts(cls):
        print("Total accounts:", cls.total_accounts)

    @classmethod
    def change_bank_name(cls, name):
        cls.bank_name = name

account1 = Bank("Rama", "12345")
account2 = Bank("Ravi", "12346")

Bank.change_bank_name("HDFC")
print(Bank.bank_name)

Bank.show_total_accounts()

```

```python
#static Method does not depend on instance variables or class variables.
class Bank:

    @staticmethod
    def validate_amount(amount):
        if amount > 0:
            return True
        else:
            return False

print(Bank.validate_amount(1500))
print(Bank.validate_amount(-500))
```

```python
class Student:

    # Class variable
    school_name = "DAV School"

    # Constructor
    def __init__(self, name, rollno, marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks

    # Instance method
    def display(self):
        print("Student Name:", self.name)
        print("Roll Number:", self.rollno)
        print("Marks:", self.marks)
        print("School:", Student.school_name)

    # Instance method
    def check_result(self):
        if self.marks >= 35:
            print("Result: Pass")
        else:
            print("Result: Fail")

    

    def validate_marks(marks):
        if 0 <= marks <= 100:
            return True
        else:
            return False


# Creating objects
student1 = Student("Rama", 101, 85)
student2 = Student("Ravi", 102, 30)

# Displaying student details
student1.display()
student1.check_result()

student2.display()
student2.check_result()

# Calling static method
print("\nMarks validation:")
print(Student.validate_marks(85))
print(Student.validate_marks(150))

```

```python
class Employee:
    company_name = "XYZ Solutions"

    def __init__(self, name, emp_id, basic_salary):
        self.name = name
        self.emp_id = emp_id
        self.__basic_salary = basic_salary

    def get_salary(self):
        return self.__basic_salary

    def calculate_salary(self):
        hra = self.__basic_salary * 0.20
        da = self.__basic_salary * 0.10

        final_salary = self.__basic_salary + hra + da

        return hra, da, final_salary

    def display_details(self):
        hra, da, final_salary = self.calculate_salary()

        print("\nEmployee Name:", self.name)
        print("Employee ID:", self.emp_id)
        print("Company:", Employee.company_name)
        print("Basic Salary:", self.__basic_salary)
        print("HRA:", hra)
        print("DA:", da)
        print("Final Salary:", final_salary)

    def validate_salary(amount):
        return amount > 0

    def __del__(self):
        print("Employee record deleted:", self.name)


emp1 = Employee("Rama", 101, 30000)
emp2 = Employee("Ravi", 102, 40000)
emp3 = Employee("Priya", 103, 50000)



emp1.display_details()

emp2.display_details()

emp3.display_details()

print("\nBasic Salary of Rama:", emp1.get_salary())

print("\nSalary Validation:")
print(Employee.validate_salary(25000))
print(Employee.validate_salary(-500))

del emp1
del emp2
del emp3

```

```python
from abc import ABC, abstractmethod


# Abstract parent class
class Student(ABC):

    @abstractmethod
    def perform_work(self):
        pass
    def enrolled(self):
        print("Student of NSTI")

# Child class implements the required method
class AIPA(Student):

    def perform_work(self):
        print("Studying AI")

# Child class implements the required method
class CSA(Student):

    def perform_work(self):
        print("Fieldwork")

s1 = AIPA()
s2 = CSA()

s1.perform_work()
s2.perform_work()
s1.enrolled()
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self,Name,age):
         self.Name= Name 
         self.age = age
    def display(self):
        print(self.Name, self.age, end=" ") 
        self.role()
    
    @abstractmethod
    def role(self):
         pass
class Manager(Employee):
    def role(self):
         print("manage projects")
class Developer(Manager):
    def role(self):
         print("works with manager")

e1=Manager("Ram",30)
e2=Developer("Hari",40)

e1.display()
e2.display()
          
```
